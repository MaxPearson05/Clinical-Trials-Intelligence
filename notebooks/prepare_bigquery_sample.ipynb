# Prepare the API sample for BigQuery

Converts the API extract to newline-delimited JSON, with one study per line. Run the ingestion notebook first to create data/raw/clinicaltrials_project_dev_200.json.

The original raw JSON for the archived 200-study sample is not included, so a new extraction may produce different records. The recorded 200-row validation describes the historical sample. See [sources and validation](../docs/sources-and-validation.md).


```python
import json
from pathlib import Path

source_path = Path("../data/raw/clinicaltrials_project_dev_200.json")
output_path = Path("../data/raw/clinicaltrials_project_dev_200.ndjson")

with open(source_path, "r", encoding="utf-8") as file:
    raw_data = json.load(file)

with open(output_path, "w", encoding="utf-8") as file:
    for study in raw_data["studies"]:
        file.write(json.dumps(study) + "\n")

print("Studies written:", len(raw_data["studies"]))
print("File created:", output_path.exists())
print("Output:", output_path)
```

## Preparing JSON for BigQuery

The ClinicalTrials.gov API extract was originally stored as one JSON object containing a nested `studies` array.

For BigQuery ingestion, I converted the file to newline-delimited JSON (NDJSON), where each line represents one complete clinical trial record.

The conversion:
- Reads the original API JSON
- Extracts each study from the `studies` array
- Writes one study per line
- Preserves nested structures such as interventions and locations

Validation:
- Source studies: 200
- NDJSON records written: 200

This creates a BigQuery-compatible input while retaining the nested data required for later `UNNEST` analysis.
